## The Ames Housing Dataset

The [Ames Housing dataset](http://jse.amstat.org/v19n3/decock.pdf) is a newer (2011) replacement for the classic Boston Housing dataset. Each record represents a residential property sale in Ames, Iowa. It contains many different potential predictors and the target variable is `SalePrice`.

````python
import pandas as pd
ames_df = pd.read_csv("ames.csv", index_col=0)
ames_df
````

````python
ames_df.describe()
````

We will focus specifically on a subset of the overall dataset. These features are:

```
LotArea: Lot size in square feet

1stFlrSF: First Floor square feet

GrLivArea: Above grade (ground) living area square feet
```

````python
ames_subset = ames_df[['LotArea', '1stFlrSF', 'GrLivArea', 'SalePrice']].copy()
ames_subset
````

## Step 1: Visualize Relationships Between Features and Target

For each feature in the subset, create a scatter plot that shows the feature on the x-axis and `SalePrice` on the y-axis.

````python
# Your code here - import relevant library, create scatter plots
import matplotlib.pyplot as plt 
import seaborn as sns  

features = ['LotArea', '1stFlrSF', 'GrLivArea']
for feature in features:
    plt.figure(figsize=(10,6))
    sns.scatterplot(data=ames_df, x=feature, y='SalePrice')
    plt.xlabel(feature)
    plt.ylabel('SalePrice')
    plt.grid(True)
    plt.show()
````

````python
# Your written answer here - do these seem like good candidates for linear regression?

# 1. LotArea - a bigger lot area does not necessarily mean that the price of the house will be higher. The relationship between the 
# lot area and sale price is weak and therefore this feature is not a good candidate for linear regression.
# 2. 1stFlrSF - First floor square feet shows a positive linear relationship because from the scatter plot we see that more square feet
# lead to high sales price of the house. This feature is therefore a good candidate for linear regression.
# 3. GrLivArea - This feature shows a strong relationship with the sale price because as the square feet increase the sales price 
# increases and would recommend that it is a good candidate for linear regression. 


````

## Step 2: Build a Simple Linear Regression Model

Set the dependent variable (`y`) to be the `SalePrice`, then choose one of the features shown in the subset above to be the baseline independent variable (`X`).

Build a linear regression using StatsModels, describe the overall model performance, and interpret its coefficients.

````python
# Your code here - define y and baseline X

X_baseline = ames_df['GrLivArea']
y = ames_df['SalePrice']

````

````python
# Your code here - import StatsModels, fit baseline model, display results
import statsmodels.api as sm 


baseline_model = sm.OLS(y, sm.add_constant(X_baseline)).fit()
print(baseline_model.summary())
````

````python
# Your written answer here - interpret model results

# The R-squared os 0.502 which means that 50% of the variability in sale price is attributed to the grade living area square feet. 
# In addition, for every increase in a square foot, the sale price increases by $107 which assumes a linear relationship.
# The P-value is 0 which tells us that the model is statistically significant.
````

## Step 3: Build a Multiple Linear Regression Model

For this model, use **all of** the features in `ames_subset`.

````python
# Your code here - define X
X = ames_df[['LotArea', '1stFlrSF', 'GrLivArea']]
````

````python
# Your code here - fit model and display results

fit_model = sm.OLS(y, sm.add_constant(X)).fit()
print(fit_model.summary())
````

````python
# Your written answer here - interpret model results. Does this model seem better than the previous one?

# This model is better than the previous one because it explains more variance of 56% as opposed to 50% in the previous one. The model
# is also still statistically significant overall.
````

## Step 4: Create Partial Regression Plots for Features

Using your model from Step 3, visualize each of the features using partial regression plots.

````python
# Your code here - create partial regression plots for each predictor
from statsmodels.graphics.regressionplots import plot_partregress_grid

fig = plt.figure(figsize=(12, 8))
plot_partregress_grid(fit_model, fig=fig)
plt.tight_layout()
plt.show()
````

````python
# Your written answer here - explain what you see, and how this relates
# to what you saw in Step 1. What do you notice?

# GrLivArea still shows a strong positive linear trend because as the square feet increase, the sale price increases as well.
# 1stFlrSF also shows a positive linear trend however it is more spread out which shows less correlation in the partial regression.
# The LotArea partial regression plot shows a weak relationship with saleprice just like it did in the scatter plots.
# In conclusion, the scatter plot and the partial regression plot have given the same kind of results. GrLivArea is a strong 
# predictor, 1stFlrSF a moderate predictor while the LotArea is a weak predictor.
````

## Level Up (Optional)

Re-create this model in scikit-learn, and check if you get the same R-Squared and coefficients.

````python
# Your code here - import linear regression from scikit-learn and create and fit model

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
````

````python
# Your code here - compare R-Squared

linear_reg = LinearRegression()
linear_reg.fit(X,y)
y_pred = linear_reg.predict(X)
r2 = r2_score(y, y_pred)
print(r2)
````

````python
# Your code here - compare intercept and coefficients

print(linear_reg.intercept_)
print(linear_reg.coef_)
````
